```python
from scripts.parte1_pd.entrenar_pd import entrenar, imprimir_resumen
from src.configuracion import cargar_config
```

## Red neuronal

Estos son los valores del modelo final.

```python
config = cargar_config("config.ini", "pd")

config["semilla"] = 42          # fija inicialización, orden de los lotes y dropout
config["red"].update(
    capas_ocultas=[64, 32],     # neuronas por capa oculta
    activacion="relu",          # relu | tanh | sigmoide
    l2=0.0014,                  # penalización L2; 0 = sin penalización
    dropout=0.3,                # probabilidad de apagar cada neurona; 0 = sin dropout
)
config["entrenamiento"].update(
    optimizador="adam",         # adam | sgd
    tasa=0.0026,                # tasa de aprendizaje
    epocas=200,                 # máximo de épocas
    tam_lote=512,               # tamaño del mini-lote
    paciencia=10,               # épocas sin mejorar antes de parar
)

_, _, _, metadata = entrenar(config, verbose=False)
imprimir_resumen(metadata)
```

## Regresión logística

La misma red sin capas ocultas ni dropout.

```python
config = cargar_config("config.ini", "pd")

config["semilla"] = 42          # fija inicialización, orden de los lotes y dropout
config["red"].update(
    capas_ocultas=[],           # sin capas ocultas = regresión logística
    l2=0.001,                   # penalización L2; 0 = sin penalización
    dropout=0,                  # sin neuronas ocultas no hay dropout
)
config["entrenamiento"].update(
    optimizador="adam",         # adam | sgd
    tasa=0.0026,                # tasa de aprendizaje
    epocas=200,                 # máximo de épocas
    tam_lote=512,               # tamaño del mini-lote
    paciencia=10,               # épocas sin mejorar antes de parar
)

_, _, _, metadata = entrenar(config, verbose=False)
imprimir_resumen(metadata)
```
